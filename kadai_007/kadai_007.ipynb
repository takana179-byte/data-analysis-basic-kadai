```python
# Step 1: Install and import required libraries
!pip install japanize-matplotlib

import matplotlib.pyplot as plt
import pandas as pd
import japanize_matplotlib

# --- Re-creating the merged DataFrame from Chapter 7.4 ---
# 1. Main DataFrame (sample_pandas_6)
data_main = {
    '発注日': ['2016-03-06', '2015-03-27', '2017-05-31', '2022-10-26', '2016-06-10'],
    '商品番号': ['YY4HAAZR', 'Z4WOOIYV', 'YY4HAAZR', '1QJFO8QY', '1QJFO8QY'],
    '商品名': ['商品サンプル YY4HAAZR', '商品サンプル Z4WOOIYV', '商品サンプル YY4HAAZR', '商品サンプル 1QJFO8QY', '商品サンプル 1QJFO8QY'],
    '単価': [780, 90, 780, 600, 600],
    '在庫': [20, 45, 10, 5, 40],
    '注文数': [24, 12, 0, 24, 24]
}
df = pd.DataFrame(data_main)

# 2. Category DataFrame (category.csv)
data_category = {
    '商品番号': ['YY4HAAZR', 'Z4WOOIYV', '1QJFO8QY'],
    'カテゴリー': ['弁当', 'デザート', '飲料']
}
category_df = pd.DataFrame(data_category)

# 3. Merge data (Chapter 7.4 step)
df = pd.merge(df, category_df[['商品番号', 'カテゴリー']], how='inner', on='商品番号')


# --- Step 2: Plot category frequency as a bar chart ---
df['カテゴリー'].value_counts().plot(kind='bar')

# --- Step 3: Display the chart ---
plt.show()

# --- Step 4: Calculate basic statistics of '注文数' grouped by '商品番号' ---
df.groupby('商品番号')['注文数'].describe()
```
