```python
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Create a 2x3 grid of subplots (6 regions total) with tight_layout
fig, axes = plt.subplots(2, 3, tight_layout=True)

# Step 3: Plot data into each specific subplot region

# 1. Line plot (top-left: region [0, 0])
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)

# 2. Sine wave (top-middle: region [0, 1])
x1 = np.linspace(0, 10, 100)
y1 = np.sin(x1) + 2
axes[0, 1].plot(x1, y1)

# 3. Histogram (top-right: region [0, 2])
data = np.random.normal(15, 5, 1000)
axes[0, 2].hist(data)

# 4. Scatter plot (bottom-left: region [1, 0])
np.random.seed(0)
x3 = np.random.randn(50) * 2 + 3
y3 = np.random.randn(50) * 2 + 4
axes[1, 0].scatter(x3, y3)

# Note: Regions [1, 1] and [1, 2] are left blank as permitted by the assignment.

# Step 4: Display the figure
plt.show()
```
