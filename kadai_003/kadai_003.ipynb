```python
import numpy as np

# Step 2: Define matrices A and B as multi-dimensional arrays
A = np.array([[0, 1], [2, 3]])
B = np.array([[4, 5], [6, 7]])

# Step 3: Calculate the matrix product of A and B
C = np.dot(A, B)

# Step 4: Display the matrix product result
print("Matrix Product (A @ B):")
print(C)

# Step 5: Find and display the maximum element in the result matrix
max_value = np.max(C)
print("Maximum Element:", max_value)
```
